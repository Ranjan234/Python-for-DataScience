#### **Q7. Which month has the highest number of appointments?**

```python
data['appointment_month']= data['appointment_date'].dt.month_name()
```

```python
data.sample()
```

```python
monthly_appointments = data.groupby('appointment_month')['appointment_id'].nunique().sort_values().rename('appintments')
monthly_appointments
```

```python
# Define chronological month order
month_order = [
    'January', 'February', 'March', 'April', 'May', 'June',
    'July', 'August', 'September', 'October', 'November', 'December'
]

# Reindex the monthly appointments series using the ordered months
monthly_appointments = monthly_appointments.reindex(month_order)
monthly_appointments
```

```python
plt.figure(figsize=(9, 5))
plt.plot(monthly_appointments, marker= 'o', color = '#008080')
plt.xlabel('Month')
plt.ylabel('Number of Appointments')
plt.title('Monthly Appointment Trends')
plt.grid(True)
plt.xticks(rotation = 45)
plt.savefig('Monthly Appointments Trend')
plt.show()
```

- **Insights:**
  - In the monthly appointment trend: Most patients are taking appointments IN April, January, March, and May compared to other months. Very few patients are taking appointments in September and December. Average appointments are in June, July, August, and November.

#### **Q17. Which month generates the highest billing amount?**

```python
billing_data = data.drop_duplicates('bill_id').copy()
billing_data
```

```python
monthly_billing = (
    billing_data.groupby(
        billing_data['bill_date'].dt.to_period('M')
    )['amount'].sum()
)
print(monthly_billing)
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(x=monthly_billing.index.to_timestamp(), y=monthly_billing.values, marker='o', color='#008080')
plt.title('Monthly Billing Distribution')
plt.xlabel('Month')
plt.ylabel('Total Billing Amount')
plt.grid(True)
plt.tight_layout()
plt.savefig('Monthly Billing Distribution')
plt.show()
```

- **Insights:**
 - In revenue analysis, January, April, and June show high revenue compared to other months. Patients took the most appointments in these months. The lowest revenue in December and September.
