## Step 4: Creating ML model

```python
X = df[['year']]
y = df['income']
```

```python
X.shape
```

```python
y.shape
```

```python
# train_test_split: random_state:
```

```python
# model: mathematical algo that learns from the given dataset, to proper understand
# understand the pattern
# first custom mathematical function:
# sklearn:
```

```python
model = LinearRegression()
```

```python
# training part
model.fit(X, y)
# y = mx + c
# m = slope
# x = feature
# c = intercept : when x = 0 then y = c
# give random value m, c
# gradient descent: Minimize the loss by adjusting c, m
```

```python
y_pred = model.predict(X)
print(y_pred)
```

```python
plt.scatter(df['year'], df['income'], marker = '.')
plt.title('Yearly PCI')
plt.show()
```

```python
plt.scatter(df['year'], df['income'], marker = '.', label =  'actual data')
plt.plot(df['year'], y_pred, color = 'blue', label = 'best fit line')
plt.title('Yearly PCI')
plt.legend()
plt.show()
```

```python
# residual, loss, cost
# loss = predicted - actual
# mse, mae, rmse
```

```python
df['y_pred'] = y_pred
```

```python
df
```

```python
# mae: mean absolute error
(df['income'] - df['y_pred']).mean()
```

```python
# mean_squared_error
mae =mean_absolute_error(y, y_pred)
mae
```

```python
# mean_squared_error
mse = mean_squared_error(y, y_pred)
mse
```

```python
# root_mean_squared_error
rmse  = root_mean_squared_error(y, y_pred)
rmse
```
