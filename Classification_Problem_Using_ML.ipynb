```python
y = []
for i in range(50):
  if i <= 25:
    y.append(0)
  else:
    y.append(1)
x = range(0, 50)
line_data = np.linspace(0, 1, 50)
plt.scatter(x[:25], y[:25], color = 'r', label = 'Card not Buy')
plt.scatter(x[25:], y[25:], color = 'g', label = 'Card Buy')
plt.title ("Credit Card Buy or Not")
plt.plot(line_data, color = 'b')
plt.legend()
plt.show()
```

```python
def sigmoid_func(x):
  return(1/ (1+ np**(-x)))
  # it returns s_curve
  # it never touches 0, 1
```

```python
def sigmoid_func(x):
  return(1/ (1+ np.exp(-x)))
  # it returns s_curve
  # it never touches 0, 1

line_data = np.linspace(-10, 10, 50)
s_curve = sigmoid_func(line_data)
plt.scatter(x[:25], y[:25], color = 'r', label = 'Card not Buy')
plt.scatter(x[25:], y[25:], color = 'g', label = 'Card Buy')
plt.title ("Credit Card Buy or Not")
# plt.plot(line_data, color = 'b')
plt.plot(s_curve, color = 'y')
plt.legend()
plt.show()
```
